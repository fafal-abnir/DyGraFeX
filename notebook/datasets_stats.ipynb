```python
import pandas as pd
```

```python
def print_dataset_stats(df):
    df['label'].value_counts(normalize=True)
    num_nodes = pd.unique(df[['sender','receiver']].values.ravel())
    print(f'Number of node:{len(num_nodes)}')
    num_edges = len(df)
    print(f"Number of edges:{num_edges}")
    print(f"Number of windows:{max(df['window_index'])}")
```

## Bitcoin Alpha

```python
alpha_df = pd.read_parquet("../data/BitcoinAlpha/processed/data.parquet")
alpha_df
```

```python
print_dataset_stats(alpha_df)
```
